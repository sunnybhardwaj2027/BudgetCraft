### Agent Architecture & Features:
1. **Plan-Act ReAct Loop:** Multi-step reasoning where the agent formulates thoughts, decides tool calls, receives observations, and computes decisions.
2. **Tools Called:**
   - `add_expense(item, amount, category)`: Logs an expense and updates category balance.
   - `get_summary(category)`: Retrieves total budget, spent funds, and remaining balance.
   - `set_savings_goal(target_amount, month)` *(Group of 3 Add-on)*: Sets a savings goal target.
   - `check_affordability(expense_name, cost)`: Calculates whether a purchase/trip fits remaining budget and savings targets.
3. **Memory:** Retains state (expenses logged, savings targets, category totals) and multi-turn conversation history.

```python
# Initialize environment and import agent components
from memory import BudgetMemory
from tools import reset_memory
from agent import PersonalBudgetAgent

# Reset global memory state with a student monthly budget of $5,000
memory = reset_memory(initial_budget=5000.0)
agent = PersonalBudgetAgent(memory=memory, verbose=True)
print("✅ Personal Budget Assistant Agent initialized.")
```
